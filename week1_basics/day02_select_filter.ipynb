Day 2 — select, filter, where, column operations
🎯 Goal

Query and slice DataFrames confidently — these are the most used operations in real pipelines.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.appName("Day2").getOrCreate()

data = [
    ("Ravi",    "Engineering", "Pune",      55000, "M", 28),
    ("Priya",   "HR",          "Mumbai",    42000, "F", 32),
    ("Arjun",   "Engineering", "Delhi",     72000, "M", 26),
    ("Sneha",   "Finance",     "Pune",      61000, "F", 30),
    ("Rohit",   "Engineering", "Mumbai",    80000, "M", 35),
    ("Meera",   "HR",          "Bangalore", 39000, "F", 27),
    ("Karan",   "Finance",     "Delhi",     55000, "M", 29),
    ("Divya",   "Engineering", "Pune",      91000, "F", 33),
    ("Nitin",   "HR",          "Mumbai",    44000, "M", 31),
    ("Anjali",  "Finance",     "Bangalore", 67000, "F", 28),
]

cols = ["name", "dept", "city", "salary", "gender", "age"]
df = spark.createDataFrame(data, cols)
df.show()
```

```python
# 1a. Select only name and salary
df.select("name", "salary").show()

# 1b. Select using col() object — preferred in production
df.select(col("name"), col("dept"), col("salary")).show()

# 1c. Select all columns (rarely needed but good to know)
df.select("*").show()

# 1d. Select with alias — rename column on the fly
df.select(
    col("name").alias("employee_name"),
    col("salary").alias("monthly_salary")
).show()
```

```python
df.printSchema()
```

```python
df.filter(col("salary")>60000).show()
```

```python
df.where(col("salary")>60000).show()
```

```python
df.filter("salary > 60000").show()
```

TASK 2 — filter and where
"column_name"	      Simple selections
col("column_name")	  When applying functions or conditions


Both filter() and where() do the same thing — pick either

# 2a. Single condition
df.filter(col("salary") > 60000).show()
df.where(col("salary") > 60000).show()   # same result

# 2b. String-style filter (SQL-like)
df.filter("salary > 60000").show()

# 2c. Multiple conditions — AND
df.filter((col("dept") == "Engineering") & (col("salary") > 60000)).show()

# 2d. Multiple conditions — OR
df.filter((col("city") == "Pune") | (col("city") == "Mumbai")).show()

# 2e. NOT condition
df.filter(col("dept") != "HR").show()
df.filter(~(col("dept") == "HR")).show()   # same using tilde

# 2f. Filter on string match
df.filter(col("city") == "Pune").show()

# 2g. isin() — filter multiple values
df.filter(col("city").isin("Pune", "Mumbai")).show()

# 2h. NOT isin()
df.filter(~col("city").isin("Pune", "Mumbai")).show()

```python
from pyspark.sql.functions import col, lit, upper, lower, length

# 3a. Add 10% bonus to salary as new column
df.select(
    col("name"),
    col("salary"),
    (col("salary") * 1.10).alias("salary_with_bonus")
).show()

# 3b. String functions on column
df.select(
    upper(col("name")).alias("name_upper"),
    lower(col("city")).alias("city_lower"),
    length(col("name")).alias("name_length")
).show()

# 3c. Add a literal constant column
df.select(
    col("name"),
    col("salary"),
    lit("India").alias("country")
).show()
```

```python
# Scenario: Get name, dept, salary of Engineering employees
# in Pune earning more than 60000

result = df \
    .filter(col("dept") == "Engineering") \
    .filter(col("city") == "Pune") \
    .filter(col("salary") > 60000) \
    .select("name", "dept", "salary","city")

result.show()
```
