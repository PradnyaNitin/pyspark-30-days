from pyspark.sql.functions import col, asc, desc

# 1a. Sort by salary ascending (default)
df.orderBy("salary").show()

# 1b. Sort descending
df.orderBy(col("salary").desc()).show()
df.orderBy(desc("salary")).show()          # same thing

# 1c. Sort by multiple columns
df.orderBy(col("dept").asc(), col("salary").desc()).show()

# 1d. sort() is alias for orderBy()
df.sort(col("salary").desc()).show()

# 1e. Real world — top 3 earners
df.orderBy(col("salary").desc()).limit(3).show()

# 1f. Sort after groupBy — department wise sorted salary
df.groupBy("dept") \
  .avg("salary") \
  .orderBy(col("avg(salary)").desc()) \
  .show()

```python
# 2a. Get top 5 rows
df.limit(5).show()

# 2b. Combine with orderBy for Top-N pattern
# Top 5 highest paid employees
df.orderBy(col("salary").desc()).limit(5).show()

# Bottom 3 earners
df.orderBy(col("salary").asc()).limit(3).show()

# 2c. Top earner per department
# (preview — full solution uses Window functions on Day 10)
df.orderBy(col("salary").desc()) \
  .groupBy("dept") \
  .agg({"salary": "max"}) \
  .show()
```

```python
# 3a. Distinct rows (entire row must be duplicate)
df.distinct().show()
print("Before distinct:", df.count())
print("After distinct: ", df.distinct().count())

# 3b. Distinct values in one column
df.select("dept").distinct().show()
df.select("city").distinct().show()

# 3c. Distinct on multiple columns
df.select("dept", "city").distinct().show()
```

```python
# dropDuplicates is more powerful than distinct
# because you can control WHICH columns define a duplicate

# 4a. Drop full row duplicates (same as distinct)
df.dropDuplicates().show()

# 4b. Drop duplicates based on specific columns
# Keep first occurrence of each name
df.dropDuplicates(["name"]).show()

# 4c. Drop duplicates based on name + dept combination
df.dropDuplicates(["name", "dept"]).show()

# 4d. Real world — deduplicate incoming data before loading
df_clean = df.dropDuplicates(["name", "dept", "salary"])
print("Cleaned row count:", df_clean.count())
display(df_clean)
```

```python
# Business Ask:
# "Give me top 3 unique employees per city
#  sorted by salary, no duplicate names"

result = df \
    .dropDuplicates(["name"]) \
    .orderBy(col("salary").desc()) \
    .select("name", "city", "dept", "salary") \
    .limit(10)

display(result)
```
