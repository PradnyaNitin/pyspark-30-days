```python
# From Claude self learning

from pyspark.sql import SparkSession

# This is the entry point to ALL Spark operations
spark = SparkSession.builder.appName("Day1_Practice").master("local[*]").getOrCreate()

# Check it works
print(spark.version)
print(spark.sparkContext.appName)
```

```python
data = [
    ("Alice", 28, "Mumbai"),
    ("Bob", 32, "Pune"),
    ("Charlie", 25, "Delhi"),
    ("Diana", 29, "Bangalore"),
    ("Eve", 35, "Mumbai")
]
```

```python
columns = ["name", "age", "city"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
# Always do these 5 things first with any new dataset

# 1. See the data
df.show()          # default 20 rows
df.show(3)         # first 3 rows
df.show(truncate=False)  # show full column values

# 2. See the structure
df.printSchema()   # column names + data types

# 3. Check size
print(df.count())              # row count
print(len(df.columns))         # column count

# 4. Summary statistics
df.describe().show()           # count, mean, std, min, max

# 5. See column names
print(df.columns)
print(df.dtypes)
```
